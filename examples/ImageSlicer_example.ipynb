```python
import sys, os
import numpy as np
from skimage import data
import matplotlib.pyplot as plt

# Add parent directory to the Python path
notebook_dir = os.getcwd()
parent_dir = os.path.abspath(os.path.join(notebook_dir, '..'))
sys.path.append(parent_dir)

# Import the DFT registrators form the py_tools package
from impro_tools.ImageSlicer import ImageSlicer
```

```python
img = data.astronaut()
slicer = ImageSlicer()

# Example usage of slice_along_line
p1 = (150, 100)
p2 = (400, 300)
thickness = 50
slice_img, rect = slicer.slice_along_line(img, p1, p2, thickness)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img)
plt.plot([p1[0], p2[0]], [p1[1], p2[1]], color='red')
plt.gca().add_patch(plt.Polygon(rect[:, :2], closed=True, fill=False, edgecolor='blue', linewidth=2))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Slice Along Line")
plt.imshow(slice_img)
plt.axis('off')

plt.show()
```

```python
# Load another image for the next example
img = data.logo()

# Example usage of slice_by_corners
p1 = (150, 100)
p2 = (300, 150)
thickness = 50
slice_img = slicer.slice_by_corners(img, p1, p2)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img)
plt.gca().add_patch(plt.Polygon([p1, (p2[0], p1[1]), p2, (p1[0], p2[1])], closed=True, fill=False, edgecolor='blue', linewidth=2))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Slice Along a Rectangle")
plt.imshow(slice_img)
plt.axis('off')

plt.show()
```
